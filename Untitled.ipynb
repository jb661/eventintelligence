```python
import os
import time
import requests
import pandas as pd
from dotenv import load_dotenv

load_dotenv()
API_KEY = os.getenv("API_KEY")

def get_event_inventory_status(event_id: str) -> dict:
    """Fetch availability and dynamic price ranges for a specific event ID."""
    url = "https://app.ticketmaster.com/inventory-status/v1/availability"
    params = {
        "events": event_id,
        "apikey": API_KEY
    }
    
    try:
        response = requests.get(url, params=params, timeout=10)
        response.raise_for_status()
        return response.json()  # Converts JSON response into a Python dictionary natively
    except requests.exceptions.RequestException as e:
        print(f"Error fetching inventory for event {event_id}: {e}")
        return {}

if __name__ == "__main__":
    # Load raw events using pandas
    df = pd.read_json("data/raw/events_2026-08-06.json")

    # Iterate through the first 5 events
    for _, row in df.head(5).iterrows():
        event_id = row.get("id")
        event_name = row.get("name")
        
        print(f"Checking inventory for: {event_name} (ID: {event_id})")
        inventory = get_event_inventory_status(event_id)
        
        print(inventory)
        print("-" * 50)
        
        # Rate limiting pause (5 requests/sec max)
        time.sleep(0.25)
```

```python
def parse_price_field(row, field_name):
    # Check if priceRanges exists and is a non-empty list
    price_list = row.get("priceRanges")
    if isinstance(price_list, list) and len(price_list) > 0:
        return price_list[0].get(field_name)
    return None

# Use .apply across rows to safely handle missing keys
df["min_price"] = df.apply(lambda row: parse_price_field(row, "min"), axis=1)
df["max_price"] = df.apply(lambda row: parse_price_field(row, "max"), axis=1)
df["currency"] = df.apply(lambda row: parse_price_field(row, "currency"), axis=1)

# Display results
print(df[["name", "min_price", "max_price", "currency"]].head())
```

To tackle the small sample problem, we want to introduce a smoothing weight $W$ that will give a better *smoothed* LQ quotient.
The formula to compute it is as follows:
$$
\frac{C_g+WN}{CN +WN}
$$
where $C_g$ is the count of events of genre $g$ in the city and $C = \sum_g C_g$ the total per city
